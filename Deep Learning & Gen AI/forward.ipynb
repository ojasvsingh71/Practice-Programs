```python
import numpy as np

w13=0.5
w14=0.2
w23=0.3
w24=0.5
w35=0.1
w45=0.3

b3=0.6
b4=0.4
b5=0.8
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def mse_loss(actual, predicted):
    return np.mean((actual - predicted) ** 2)


def binary_cross_entropy(actual, predicted):
    epsilon = 1e-15
    predicted = np.clip(predicted, epsilon, 1 - epsilon)
    return -np.mean(
        actual * np.log(predicted) + (1 - actual) * np.log(1 - predicted)
    )
```

```python
# Tuple 1 (1,1,0)

x1=1
x2=1
x5_actual=0


# Tuple 2 (0,1,1)

# x1=0
# x2=1
# x5_actual=1
```

```python

def train(w13,w14,w23,w24,w35,w45,b3,b4,b5,count):

    # Forward propogation
    
    x3=sigmoid(x1*w13+x2*w23+b3)
    x4=sigmoid(x1*w14+x2*w24+b4)

    x5_pred=sigmoid(x3*w35+x4*w45+b5)

    print("Predicted value of x5 : ",x5_pred)

    
    # Loss

    mse = mse_loss(x5_actual, x5_pred)
    cross_entropy = binary_cross_entropy(x5_actual, x5_pred)
    print("MSE loss : ", mse)
    print("Binary cross-entropy loss : ", cross_entropy)


    # Error

    err=x5_actual-x5_pred
    print("Error : ",err)
    print()
```

```python
count=0
train(w13,w14,w23,w24,w35,w45,b3,b4,b5,count)
```
